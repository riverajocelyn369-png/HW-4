```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
penguins = sns.load_dataset("penguins")

penguins.head()                        
```

```python
class PenguinAnalysis:

    def __init__(self, data):
        self.data = data

        print("Dataset Shape:", self.data.shape)
        print("Column Names:", self.data.columns.tolist())

    # Method 1: Show summary statistics for numerical columns 
    def summary_stats(self):
        return self.data.describe()
        
     # Method 2: Find the average body mass for each species
    def mean_by_species(self):
        return self.data.groupby("species")["body_mass_g"].mean()
        
     # Method 3: Create a scatterplot of flipper length and body mass     
    def plot_scatter(self):
        sns.scatterplot(data=self.data, x="flipper_length_mm", y="body_mass_g")
        plt.show()
        
     # Method 4: Compare body mass across penguin species    
    def plot_boxplot(self):
        sns.boxplot(data=self.data, x="species", y="body_mass_g")
        plt.show()
        
     # Method 5: Find the correlation between flipper length and body mass
    def correlation(self):
        return self.data [["flipper_length_mm","body_mass_g"]].corr()
```

```python
analysis = PenguinAnalysis(penguins)
```

# Analysis 1: Summary Statistics
In this analysis, I want to find out the average body mass, bill length, and flipper length of the penguins. I also want to see the minimum and maximum measurements in the dataset.

```python
analysis.summary_stats()
```

# Analysis 2: Average Body Mass by Species
In this analysis, I want to find out the average body mass for each penguin species. I also want to see which species has the highest average body mass.

```python
analysis.mean_by_species()
```

# Analysis 3: Scatterplot
In this analysis I would like to see if there is a relationship between flipper length and body mass for penguins. I want to determine if penguins with longer flippers tend to have higher weights.

```python
analysis.plot_scatter()
```

# Analysis 4: Boxplot
In this analysis I want to compare body masses of different penguin species. I’m curious to know which species usually has the highest body mass and how their weights vary.

```python
analysis.plot_boxplot()
```

# Analysis 5: Correlation
In this analysis, I want to figure out how strongly flipper length and body mass are related in penguins. I want to see if penguins with longer flippers tend to have higher body masses.

```python
analysis.correlation()
```
